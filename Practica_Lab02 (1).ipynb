## Actividad 14. Agrupaciones

Construya `resumen_carrera` agrupando por `carrera` e incluyendo:

- promedio de `nota`;
- mediana de `nota`;
- nota máxima;
- promedio de `horas_estudio`;
- cantidad de estudiantes.

Luego construya un segundo resumen por `modalidad` con:

- promedio de `nota`;
- promedio de `asistencia`;
- cantidad de estudiantes.


````python
# Escriba su solución aquí
resumen_carrera=df_limpio.groupby("carrera").aggregate(promedio_nota=("nota", "mean"),
                                                       mediana_de_nota=("nota", "median"),
                                                       nota_máxima=("nota", "max"),
                                                       promedio_de_horas_estudio=("horas_estudio", "mean"),
                                                       cantidad_de_estudiantes=("id", "count")
                                                       )
resumen_modalidad=df_limpio.groupby("modalidad").aggregate(promedio_nota=("nota", "mean"),
                                                           promedio_asistencia=("asistencia", "mean"),
                                                           cantidad_de_estudiantes=("id", "count")
                                                        )
resumen_modalidad
````
